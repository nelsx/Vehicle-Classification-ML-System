# **FEATURE ENGINEERING**

```python
import pandas as pd
from pathlib import Path

CURRENT_DIR = Path.cwd()
STORAGE_DIR = CURRENT_DIR.parent / 'Data' / 'Processed'
file_path=STORAGE_DIR / 'Cleaned_EV_Vehicle_Specs.csv'

EV_Vehicle=pd.read_csv(file_path)
```

```python
EV_Vehicle
```

## **New Feature Engineering: Fuel Efficiency Score**

```python
Fuel_Efficiency_Score = (EV_Vehicle['City_MPG']+EV_Vehicle['Highway_MPG']+EV_Vehicle['Combined_MPG'])/3
EV_Vehicle['Fuel_Efficiency_Score']=Fuel_Efficiency_Score
Fuel_Efficiency_Score.head
#This new feature, Fuel_Efficiency_Score, is calculated by taking the average of the City MPG, Highway MPG, and Combined MPG. It provides a single metric that represents the overall fuel efficiency of the vehicle, which can be a useful feature for distinguishing between different vehicle categories, as EVs typically have higher fuel efficiency compared to ICE vehicles. A higher Fuel_Efficiency_Score indicates a more fuel-efficient vehicle, which can help the model identify EVs and hybrids more effectively.
```

## **New Feature Engineering: Power Proxy**

```python
Power=EV_Vehicle['Engine_Cylinders']*EV_Vehicle['Engine_Size_L']
EV_Vehicle['Power Proxy']=Power
Power.head
#This new feature, Power, is calculated by multiplying the number of engine cylinders by the engine size in liters. It provides an estimate of the vehicle's power output, which can be a useful feature for distinguishing between different vehicle categories, as EVs typically have lower power compared to ICE vehicles.
```

## **New Feature Engineering: Emission Efficiency**

```python
CO2_per_MPG=EV_Vehicle['CO2_Emissions_g_per_mile']/EV_Vehicle['Combined_MPG']
EV_Vehicle['CO2_per_MPG']=CO2_per_MPG
CO2_per_MPG.head
#This new feature, CO2/MPG, is calculated by dividing the CO2 emissions in grams per mile by the combined miles per gallon. It provides a measure of the vehicle's emission efficiency, which can be a useful feature for distinguishing between different vehicle categories, as EVs typically have much lower CO2 emissions relative to their fuel efficiency compared to ICE vehicles. A lower CO2/MPG value indicates a more environmentally friendly vehicle, which can help the model identify EVs and hybrids more effectively.
```

## **Saving New Features**

```python
#Saving the feature engineered data to a new CSV file
savePath=CURRENT_DIR.parent / 'Data' / 'Processed' / 'Cleaned_EV_Vehicle_Specs.csv'
EV_Vehicle.to_csv(savePath)
```
