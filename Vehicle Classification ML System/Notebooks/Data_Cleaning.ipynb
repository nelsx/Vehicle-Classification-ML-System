```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from pathlib import Path
import joblib

CURRENT_DIR = Path.cwd()
STORAGE_DIR = CURRENT_DIR.parent / 'Data' / 'Raw'
file_path=STORAGE_DIR / 'EV_vs_ICE_Vehicle_Specs_2015_2026.csv'


EV_Vehicle=pd.read_csv(file_path)
EV_Vehicle.head()
```

```python
EV_Vehicle['Vehicle_Category'].value_counts()
```

```python
#Dropping irrelevant columns
EV_Vehicle=EV_Vehicle.drop(['Make','Model','Year','Fuel_Type','Drivetrain','Transmission',], axis=1)
```

```python
#Converting categorical variable to numerical variable using LabelEncoder
le=LabelEncoder()
EV_Vehicle['Vehicle_Category']=le.fit_transform(EV_Vehicle['Vehicle_Category'])

#Dropping the CNG category due to insufficient data
EV_Vehicle=EV_Vehicle[EV_Vehicle['Vehicle_Category']!=0]
```

```python
joblib.dump(le, CURRENT_DIR.parent/'Artifacts'/'LabelEncoder_Fuel_Type.joblib')
```

```python
EV_Vehicle['Vehicle_Category'].unique()
```

```python
#Checking Correlation
corr=EV_Vehicle.corr()['Vehicle_Category']
print(corr)
#Best to work with for the model are: Fuel_Type, Transmission, Engine_Cylinders, Engine_size and Co2_Emissions
```

```python
#Saving the cleaned data to a new CSV file
savePath=CURRENT_DIR.parent / 'Data' / 'Processed' / 'Cleaned_EV_Vehicle_Specs.csv'
EV_Vehicle.to_csv(savePath)
```
